### Question 1: What are the average, median, minimum and maximum house prices per unit area?

```python
# Calculate statistics for the target column: 'Y house price of unit area'
avg_price = s['Y house price of unit area'].mean()
median_price = s['Y house price of unit area'].median()
min_price = s['Y house price of unit area'].min()
max_price = s['Y house price of unit area'].max()

print(f"Average Price per Unit Area: {avg_price:.2f}")
print(f"Median Price per Unit Area: {median_price:.2f}")
print(f"Minimum Price per Unit Area: {min_price:.2f}")
print(f"Maximum Price per Unit Area: {max_price:.2f}")
```

### Question 2: What is the average house age?

```python
avg_age = s['X2 house age'].mean()
print(f"Average House Age: {avg_age:.2f} years")
```

### Question 3: Compare average price for properties near and far from the MRT station (Threshold: 500 meters)

```python
threshold = 500

near_mrt = s[s['X3 distance to the nearest MRT station'] <= threshold]
far_mrt = s[s['X3 distance to the nearest MRT station'] > threshold]

avg_price_near = near_mrt['Y house price of unit area'].mean()
avg_price_far = far_mrt['Y house price of unit area'].mean()

print(f"Properties close to MRT (<= {threshold}m) average price: {avg_price_near:.2f}")
print(f"Properties far from MRT (> {threshold}m) average price: {avg_price_far:.2f}")
```

### Question 4: Do records with more nearby convenience stores show a different average price?
